## 3.Prompt Templates

Instead of writing a fixed prompt every time, we create a reusable prompt with variables.

Example: Explain {topic} for a {level} student.

Then we can change the values:

topic = recursion<br>
level = beginner

```python
from langchain_groq import ChatGroq

model=ChatGroq(model="openai/gpt-oss-20b")
```

                    PROMPT TEMPLATES
                           │
            ┌──────────────┴──────────────┐
            │                             │
     PromptTemplate              ChatPromptTemplate
            │                             │
       Text prompts                Message prompts
            │                             │
            └──────────────┬──────────────┘
                           │
                       Variables
                           │
                    {topic}, {level}
                           │

                    Dynamic Prompts
                           │
             ┌─────────────┴─────────────┐
             │                           │
      Few-shot prompting          Prompt composition
             │                           │
       Examples + input          Combine prompt pieces

PromptTemplate

👉 Creates one text prompt

ChatPromptTemplate

👉 Creates a set/structure of chat messages with roles

### (i) PromptTemplate

```python
from langchain_core.prompts import PromptTemplate

prompt=PromptTemplate.from_template("Explain {topic} for a {level} student.")
print(prompt)
```

```python
final_prompt=prompt.format(topic="'recursion'",level="Beginner")
print(final_prompt)
```

```python
res=model.invoke(final_prompt)
print(res.content)
```

### (ii) ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt=ChatPromptTemplate.from_messages([("system","You are a helpful {role}"),
                                              ("human","Explain {topic} to me.")
                                              ])
print(chat_prompt)                                            
```

```python
final_chatPrompt=chat_prompt.format_messages(role="programming teacher",topic="resursion")
print(final_chatPrompt)
```

```python
res=model.invoke(final_chatPrompt)
print(res.content)
```

### (iii) Message Placeholder

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder

chat_prompt=ChatPromptTemplate.from_messages([
    ("system","You are a helpful programming tutor."),
    MessagesPlaceholder("history"),
    ("human","{question}")
])
print(chat_prompt)
```

```python
# Create some conversation history
from langchain_core.messages import HumanMessage,AIMessage

history=[
    HumanMessage(content="My name is Vineetha and I'm learning Python."),
    AIMessage(content="Nice to meet you,Vineetha! Python is a great language to learn.")
]
```

```python
messages=chat_prompt.format_messages(history=history,question="What programming language am i learning")
```

```python
for mes in messages:
    print(type(mes).__name__)
    print(mes.content)
    print()
```

```python
res=model.invoke(messages)
print(res.content)
```

### (iv) shot prompting

```python
zero_shot_prompt="""
Classify the following sentence as Positive or Negative.
Sentence: I really enjoyed this movie.

Answer with only Positive or Negative.
"""
```

```python
res=model.invoke(zero_shot_prompt)
print(res.content)
```

```python
one_shot_prompt="""
Classify sentences as Positive or Negative.

Example:
Sentence: I hated this movie.
Answer: Negative

Now classify:

Sentence: I really disappointed with this movie.

Answer with only Positive or Negative.

"""
```

```python
res=model.invoke(one_shot_prompt)
print(res.content)
```

```python
few_shot_prompt = """
Classify sentences as Positive or Negative.

Examples:

Sentence: I loved this movie.
Answer: Positive

Sentence: The food was terrible.
Answer: Negative

Sentence: The service was excellent.
Answer: Positive

Sentence: I hated the experience.
Answer: Negative

Now classify:

Sentence: The product quality was amazing.

Answer with only Positive or Negative.
"""
```

```python
response = model.invoke(few_shot_prompt)
print(response.content)
```
