## 6.Output Parsers

Parser = a component that takes model output and transforms it into a desired format.

### <u>Why do we even need parsers?</u>
<pre>
Let's see what happens normally.

response = model.invoke("Explain Python in one sentence.")

print(response)

You don't simply get:

Python is a programming language...

You get an AIMessage object.
  
So if our application wants just the text, we need something to extract/convert it.

That's where a parser comes in.
</pre>

```python
from langchain_groq import ChatGroq

model=ChatGroq(model="openai/gpt-oss-20b")
```

## (i) StrOutputParser:

```python
from langchain_core.output_parsers import StrOutputParser

parser=StrOutputParser()
```

```python
res=model.invoke("Why is python famous?")
print(res.content)
print(type(res))
```

```python
result=parser.invoke(res)
print(result)
print(type(result))
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt=ChatPromptTemplate.from_template("{question}")
```

```python
# Instead of manually doing we use a chain

chain=prompt | model | parser
ress=chain.invoke("what is diff between java and python")
print(ress)
```

## (ii) JSON Parser

```python
from langchain_core.output_parsers import JsonOutputParser

json_parser=JsonOutputParser()
```

```python
prompt = """
Give information about Python in JSON format.

Return exactly these fields:
- name
- type
- use

Example format:
{
    "name": "Python",
    "type": "Programming Language",
    "use": "Software Development"
}
"""
```

```python
response = model.invoke(prompt)

print(response.content)
```

```python
res=json_parser.invoke(response)
print(type(res))
print(res)
```

```python
print(res["name"])
print(res["type"])
print(res["use"])
```
