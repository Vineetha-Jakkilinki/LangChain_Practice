## 8.Text Splitters

<pre>
Main Idea:
A large document is broken into smaller pieces called chunks so that we can efficiently find the relevant information later.

First, remember Module 7.

We had:

PDF / TXT / Website
        ↓
   Document Loader
        ↓
     Document

Now Module 8 starts:

Document
    ↓
  Splitter
    ↓
Chunk 1
Chunk 2
Chunk 3
Chunk 4

That's the whole purpose of a Text Splitter.
</pre>

Note: pip install langchain-text-splitters

```python
# create a document
from langchain_core.documents import Document

text="""
Python is a high-level programming language.
It is easy to learn and widely used in software development.Python is also extremely popular in Artificial Intelligence and Machine Learning.
Libraries such as NumPy, Pandas, Scikit-learn and PyTorch are commonly used.Machine Learning allows computers to learn patterns from data.
Deep Learning is a subset of Machine Learning that uses neural networks.RAG stands for Retrieval Augmented Generation.RAG allows an LLM to retrieve information from external documents before generating an answer.
"""

doc=Document(
    page_content=text,
    metadata={
        "source":"ai_notes.txt",
        "topic":"AI"
    }
)

print(doc.page_content)
print(doc.metadata)
```

## (i) RecursiveCharacterTextSplitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter=RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=40
)
```

```python
chunks=splitter.split_documents([doc])
print("Number of chunks:",len(chunks))
```

```python
for i, chunk in enumerate(chunks):
    print(f"\n--- CHUNK {i+1} ---")
    print(chunk.page_content)
```

```python
for i, chunk in enumerate(chunks):
    print(f"\nCHUNK {i+1}")
    print("Content:", chunk.page_content)
    print("Metadata:", chunk.metadata)
```

```python
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {len(chunk.page_content)} characters")
```

<pre>
Chunk size → controls how big the pieces are.

Chunk overlap → keeps some neighboring context.

Recursive → tries to split at sensible boundaries instead of blindly cutting text.
</pre>

```python
from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader("text.pdf")
documents=loader.load()
print("Number of pages:",len(documents))
```

```python
print(documents[3].page_content)
```

```python
print(documents[3].metadata)
```

```python
# RecursiveCharacterTextSplitter

from langchain_text_splitters import RecursiveCharacterTextSplitter

recursive_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
```

```python
chunks=recursive_splitter.split_documents(documents)
print("Number of chunks:",len(chunks))
```

```python
for i,chunk in enumerate(chunks[:5]):
    print(f"\n-- CHUNK {i+1} ---")
    print(chunk.page_content)
```

```python
for i, chunk in enumerate(chunks[:10]):
    print(f"Chunk {i+1}: {len(chunk.page_content)} characters")
```

```python
# check metadata

for i, chunk in enumerate(chunks[:5]):
    print(f"\nChunk {i+1}")
    print("Metadata:", chunk.metadata)
```

## (ii) CharacterTextSplitter

```python
from langchain_text_splitters import CharacterTextSplitter

character_splitter=CharacterTextSplitter(
    separator="\n\n",
    chunk_size=500,
    chunk_overlap=50
)
```

<pre>
Here,
separator="\n\n"

means:

Try splitting based on blank lines / paragraph boundaries.
</pre>

```python
character_chunks=character_splitter.split_documents(documents)
print("Number of chunks:",len(character_chunks))
```

```python
for i, chunk in enumerate(character_chunks[:5]):
    print(f"\n--- CHUNK {i+1} ---")
    print(chunk.page_content)
```

```python
print("RecursiveCharacterTextSplitter:")
print(len(chunks))

print("\nCharacterTextSplitter:")
print(len(character_chunks))
```

## (iii) TokenTextSplitter

```python
from langchain_text_splitters import TokenTextSplitter

token_splitter=TokenTextSplitter(chunk_size=200,chunk_overlap=20)
```

```python
token_chunks=token_splitter.split_documents(documents)
print("Number of chunks:",len(token_chunks))
```

```python
for i, chunk in enumerate(token_chunks[:5]):
    print(f"\n--- CHUNK {i+1} ---")
    print(chunk.page_content)
```

```python
print("Recursive chunks:", len(chunks))
print("Character chunks:", len(character_chunks))
print("Token chunks:", len(token_chunks))
```
