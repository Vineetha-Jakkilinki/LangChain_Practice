## 1.Direct usage using SDK vs Langchain

```python
# Direct usage using SDK
from google import genai

client=genai.Client()
response=client.models.generate_content(model="gemini-3.8-flash",contents="Explain what an LLM is in simple words.")
print(response.text)
```

```python
res=client.models.generate_content(model="gemini-3.6-flash",contents="Give me a java program to reverse a string and explain every line of code")
print(res.text)
```

```python
client=genai.Client()
res=client.models.generate_content(model="gemini-3.6-flash",
                                   contents="""
                                   Explain ML, DL and GenAI.

                                   Give the answer in this format:

                                   ML:
                                   DL:
                                   GenAI:
                                   """
                                   )
print(res.text)
```

```python
#streaming
res=client.models.generate_content_stream(model="gemini-3.6-flash",contents="Explain how a neural network works in detail.")
for chunk in res:
    print(chunk.text,end="")
```

```python
# Using langchain
from langchain_google_genai import ChatGoogleGenerativeAI

model=ChatGoogleGenerativeAI(model="gemini-3.6-flash")
response=model.invoke("EXplain what an LLM is in simple terms.")
print(response.content)
```

```python
response=model.invoke("Give me 5 important SQL questions for a fresher along with query solution")
print(response.content)
```

```python
# Streaming with langchain

for chunk in model.stream(
    "Explain how a neural network works in detail."
):
    print(chunk.content,end="")
```
