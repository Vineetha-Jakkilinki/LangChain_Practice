## 9.Embeddings

<pre>
What is an embedding?

An embedding is a way of converting text into numbers (a vector) so that a computer can compare the meaning of different pieces of text.

An embedding model takes text:

"cat"

and converts it into something like:

[0.12, -0.45, 0.87, 0.31, ...]

This is called a vector.

</pre>

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
text="Python is a programming language."
vector=embeddings.embed_query(text)
print(type(vector))
print(len(vector))
print(vector)
```

```python
texts=[
    "I have a cat",
    "I have a kitten",
    "I like databases"
]
vectors=embeddings.embed_documents(texts)
for i,vector in enumerate(vectors):
    print(f"Text {i+1}:{len(vector)} dimensions")
```

### Note: pip install scikit-learn

```python
# performing similarity search using cosine similarity
from sklearn.metrics.pairwise import cosine_similarity

similarity1=cosine_similarity([vectors[0]],[vectors[1]])
print(similarity1)
```

```python
similarity2 = cosine_similarity([vectors[0]],[vectors[2]])
print(similarity2)
```

```python
similarity3=cosine_similarity([vectors[1]],[vectors[2]])
print(similarity3)
```

<pre>
Vector 1 ↔ Vector 2  → HIGH similarity
Vector 1 ↔ Vector 3  → LOW similarity
Vector 2 ↔ Vector 3  → LOW similarity
</pre>

## Embedding models
<pre>
1. all-MiniLM-L6-v2 ⭐⭐⭐⭐⭐

From Sentence Transformers
384-dimensional embeddings
Lightweight and fast
Very good for learning, semantic search, and small/medium RAG projects
We can continue using it for our practical work.

2. BGE models ⭐⭐⭐⭐⭐

BGE = BAAI General Embedding

Examples:

BAAI/bge-small-en-v1.5
BAAI/bge-base-en-v1.5
BAAI/bge-large-en-v1.5

Why learn?

BGE models are widely used for semantic search and RAG.

Think:

MiniLM → lightweight / fast
BGE    → stronger retrieval


3. E5 models ⭐⭐⭐⭐

Examples:

intfloat/e5-small-v2
intfloat/e5-base-v2
intfloat/e5-large-v2

E5 is another popular family designed particularly for text retrieval.

You'll often see:

query: What is RAG?

## Mini-project 🛠️
Semantic Search Engine

Input:

"How do I improve database performance?"

Search documents containing:

"SQL indexing"
"query optimization"
"database tuning"

even if the exact words aren't present.


```python
from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader("text.pdf")
documents=loader.load()
print(len(documents))
```

```python
print(documents[14].page_content)
documents[14].metadata
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
text_content=[doc.page_content for doc in documents]
vectors=embeddings.embed_documents(text_content)

```

```python
query="What core institutional framework questions did the Union Constitution Committee consider, and what system of governance did they ultimately choose?"
query_embed=embeddings.embed_query(query)
query_embed
```

```python
from sklearn.metrics.pairwise import cosine_similarity

scores=cosine_similarity([query_embed],vectors)[0]
print(scores)
```

```python
results=[]

for i,score in enumerate(scores):
    results.append((documents[i],score))
results.sort(key=lambda x: x[1], reverse=True) 
```

```python
for document, score in results:
    print(f"Score: {score:.4f}")
    print(document)
    print()
```
