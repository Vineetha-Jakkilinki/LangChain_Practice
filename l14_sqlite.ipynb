Python’s built-in sqlite3 library acts as a bridge, allowing you to write pure SQL commands directly inside your Python cells.

<pre>

1. conn opens the database.
2. cursor does the work (runs the SQL text).
3. conn.commit() saves the work permanently to the disk.
4. conn.close() locks the cabinet and leaves.
</pre>

```python
# create a database and table
import sqlite3

# 1. Connect to a database (it creates 'school.db' file if it doesn't exist)
conn=sqlite3.connect('school.db')
cursor=conn.cursor()
```

```python
# 2.create a table using SQL
cursor.execute(''' 
CREATE TABLE IF NOT EXISTS student(
  id INTEGER PRIMARY KEY AUTOINCREMENT,
  name TEXT NOT NULL,
  age INTEGER,
  major TEXT
)
''')
```

```python
# 3.Save (commit) changes and close connection

conn.commit()
print("Database and 'student' table created succesfully!")
```

```python

#Insert rows into the table
cursor.execute("INSERT INTO student (name, age, major) VALUES('Alice',20,'CSE')")
cursor.execute("INSERT INTO student (name, age, major) VALUES('Bob',22,'Data science')")
cursor.execute("INSERT INTO student (name, age, major) VALUES('Charlie',21,'Physics')")

conn.commit()
print("Students added successfully!")
```

```python
# Query(Read) the data

cursor.execute("SELECT * FROM student")
rows=cursor.fetchall()

print("ID | NAME | AGE | MAJOR")
print("-"*30)
for row in rows:
    print(f"{row[0]} | {row[1]} | {row[2]} | {row[3]}")
```

<pre>
1. cursor.execute("SELECT * FROM student")
	• You tell your assistant (cursor) to go run the SQL command. The assistant runs it, gets the data, and holds it in memory.
2. rows = cursor.fetchall()
	• The assistant brings back all the rows as a giant Python list of tuples. It looks like this behind the scenes: [(1, 'Alice', 20, 'CSE'), (2, 'Bob', 22, 'Data science')].
3. for row in rows:
	• This loops through that list, one student at a time. For the first loop, row is (1, 'Alice', 20, 'CSE').
4. print(f"{row[0]} | {row[1]} ...")
	• Since row is a Python tuple, you have to print each column manually by its index number:
		• row[0] is the 1st column (id)
		• row[1] is the 2nd column (name)
		• row[2] is the 3rd column (age)
		• row[3] is the 4th column (major)
</pre>
