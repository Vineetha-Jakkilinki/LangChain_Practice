## 7.Document Loaders

<pre>
We already know:

User → Prompt → LLM → Answer

Now we're going to learn how to give the LLM information from files, PDFs, websites, CSVs, folders, etc.
</pre>

<pre>
Instead of manually writing different code for every file type, LangChain provides Document Loaders.

PDF / TXT / CSV / Website / Folder
              ↓
       Document Loader
              ↓
        Document objects
              ↓
       Later → RAG
</pre>

<pre>
What is a Document?

In LangChain, a Document is basically a container for some content plus information about where that content came from.

Think:

Document
├── page_content
└── metadata

For example, suppose we have a PDF containing:

Python is a high-level programming language.

LangChain can represent it approximately as:

Document(
    page_content="Python is a high-level programming language.",
    metadata={"source": "python.pdf"}
)

</pre>

## (i) PDF Loader

```python
from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader("Computer_Networks_Complete_Notes-Shivansh-Vasu.pdf")
documents=loader.load()
print(type(documents))
```

```python
len(documents)
```

```python
print(documents[18].page_content)
```

```python
print(documents[19].metadata)
```

## (ii) Text Loader

```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader("noun.txt",encoding="utf-8")
documents=loader.load()
print(documents[0].page_content)
print(len(documents))
```

## (iii) CSV Loader


```python
from langchain_community.document_loaders import CSVLoader

loader=CSVLoader("precautions_df.csv")
documents=loader.load()
print(documents[0])
```

## (iv) Web Loader

<pre>
Suppose the information you want is on a website.

You want LangChain to read the webpage.

Normally:

Website
   ↓
Web Loader
   ↓
Document
</pre>

NOTE: 
install pip install beautifulsoup4


```python
from langchain_community.document_loaders import WebBaseLoader

loader=WebBaseLoader("https://en.wikipedia.org/wiki/GitHub")
documents=loader.load()
print(len(documents))
```

```python
print(documents[0].page_content[:500])
```

## (v) Directory Loader

```python
from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import TextLoader

loader=DirectoryLoader(
    "my_documents",
    glob="*.txt",
    loader_cls=TextLoader
)
documents=loader.load()
print(len(documents))
```

```python
documents[2].page_content
```

```python
documents[2].metadata
```
