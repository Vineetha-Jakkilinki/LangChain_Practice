### The four message types:<br>
SystemMessage: AI behavior/instructions<br>
HumanMessage: User input<br>
AIMessage: Previous AI response<br>
ToolMessage: Tool result

```python
from langchain_core.messages import SystemMessage
from langchain_core.messages import HumanMessage
from langchain_core.messages import AIMessage

sys_msg=SystemMessage(content="You are a helpful Python tutor.")
hum_msg=HumanMessage(content="EXplain python generator concept in simple words")
messages=[sys_msg,hum_msg]
res=model.invoke(messages)
res.content
```

```python
# Sending previous history
messages=[
    SystemMessage(content="Bro You are a helpul programming tutor"),
    HumanMessage(content="My name is Vineetha"),
    AIMessage(content="Nice to meet you,Vineetha"),
    HumanMessage(content="What is the difference between ai and ml?")
]
res=model.invoke(messages)
print(res.content)
```

```python
messages.append(res)
messages.append(HumanMessage(content="What is generative ai and how it is different from ai and ml?"))
response=model.invoke(messages)
```

## TASK

Role-Based AI Assistant

Build:

Teacher Mode,
Coder Mode,
Interviewer Mode,
Friend Mode

using system messages.

```python
from langchain_groq import ChatGroq

model=ChatGroq(model="openai/gpt-oss-20b")
```

```python
n=int(input("Enter choice:"))
sys_msg=""
match n:
    case 1:
      sys_msg="You are a teacher.Explain concepts clearly.Assume the student is a beginner.Break difficult concepts into smaller parts.Give examples.Avoid unnecessarily complicated terminology.Ask/check whether the student understood when appropriate." 
    case 2: 
      sys_msg="You are a coding mentor and you should focus on:programming logic,correct code,time complexity,edge cases,debugging,practical implementation."
    case 3:
      sys_msg="You are a interviewer.Ask one question at a time.Do not immediately reveal the answer.Wait for the candidate's response.Evaluate the response and provide feedback."
    case 4:
      print("Friend Mode")
      sys_msg="Be casual.Use friendly language.Keep explanations conversational.Encourage the user.Don't sound overly formal."
   
```

```python
from langchain_core.messages import SystemMessage,HumanMessage

sys_mes=SystemMessage(content=sys_msg)
match n:
    case 1:
      res1 = model.invoke([sys_msg, HumanMessage(content="Explain recursion.")])
      print("Teacher mode:")
      print(res1.content)
    case 2:
      res2=model.invoke([sys_mes,HumanMessage(content="Explain recursion.")])
      print("Coder mode:")
      print(res2.content) 
    case 3:
      res3=model.invoke([sys_mes,HumanMessage(content="Start my interview on SQL.")])
      print("Interviewer mode:")
      print(res3.content)
    case 4:
      res4=model.invoke([sys_mes,HumanMessage(content="Bro, I'm struggling with recursion.")])
      print("Friend mode:")
      print(res4.content)
```
